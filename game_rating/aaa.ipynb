```python
def normalize_path(path):
    rep_path = path
    while "\\" in rep_path or "//" in rep_path:
        rep_path = rep_path.replace('\\',"/").replace('//','/')
    return rep_path


# normalize_path("\\\\\d\\\s\\\s\\\\\s")
normalize_path("/////a///")
```
